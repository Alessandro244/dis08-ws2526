```python
spam_1 = ['apples', 'bananas', 'tofu', 'cats']
spam_2 = ['apples', 'bananas', 'tofu', 'cats', 'dogs', 'fox']
spam_3 = ['bananas']
spam_4 = []

def listtostr(liste):
    
    string = ""

    if len(liste) == 0: # leere Ausgabe bei leerer Liste
        return ""
    
    if len(liste) == 1: # nur ein Element → direkt zurückgeben
        return liste[0]
        
    for i in range(len(liste)):
        if i == 0: # erstes Element der Liste
            string = liste[i]
        elif i == len(liste) - 1: # letztes Element der Liste vor dem ein "and" stehen soll
            string = string + ", and " + liste[i]
        else:
            string = string + ", " + liste[i]
    return string

print(listtostr(spam_1)) # Funktion testen
# print(listtostr(spam_2)) # Funktion testen, wenn Liste mehr als vier Elemente hat
# print(listtostr(spam_3)) # Funktion testen, wenn Liste nur ein Element hat
# print(listtostr(spam_4)) # Funktion testen, wenn Liste keine Element hat
```

# Coin Flip Streak

```python
import random

numberOfStreaks = 0

for experimentNumber in range(10000):
    # Code that creates a list of 100 'heads' or 'tails' values.
    list_random = []
    
    for i in range(100):
        if random.randint(0, 1) == 0:
            list_random.append("H")
        else:
            list_random.append("T")
            
    # Code that checks if there is a streak of 6 heads or tails in a row.
    counter_T = 0
    counter_H = 0
    
    for i in list_random:
        if i == "T":
            counter_T = counter_T + 1
            if counter_T == 6:
                numberOfStreaks = numberOfStreaks + 1
                counter_T = 0 # Setz counter wieder zurück
                break
        else:
            counter_T = 0 # Nicht 'T' nacheinander -> Counter zurücksetzen
            
        if i == "H":
            counter_H = counter_H + 1
            if counter_H == 6:
                numberOfStreaks = numberOfStreaks + 1
                counter_H = 0 # Setz counter wieder zurück
                break
        else:
            counter_H = 0 # Nicht 'H' nacheinander -> Counter zurücksetzen
            
    continue

print('Chance of streak: %s%%' % (numberOfStreaks / 100))
```

# Fantasy Game Inventory

```python
stuff_1 = {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12}
stuff_2 = {'water bottle': 3, 'map': 1, 'apple': 10, 'bow': 1, 'silver coin': 30}

def displayInventory(inventory):
    print("Inventory:")
    
    item_total = 0
    
    for k, v in inventory.items():
        # FILL THIS PART IN
        print(inventory[k], k) # Keys und Values printen
        
        item_total = item_total + v # Alle Items zusammenaddieren
    print("Total number of items: " + str(item_total))

displayInventory(stuff_1) # Test
#displayInventory(stuff_2) # Test mit anderem Dictionary 
```

# List to Dictionary Function for Fantasy Game


```python
dragonLoot = ['gold coin', 'dagger', 'gold coin', 'gold coin', 'ruby']
inv = {'gold coin': 42, 'rope': 1}

def addToInventory(inventory, addedItems):
    for item in addedItems:
        if item in inventory:
            inventory[item] = inv[item] + 1
        else:
            inventory[item] = 1
            
    return inventory

inv = addToInventory(inv, dragonLoot)
displayInventory(inv)
```

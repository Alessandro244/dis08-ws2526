```python
import re

def main():
    
    # CSV Datei einlesen
    with open("orders.csv") as f_in:
        text = f_in.read()
        zeilen = text.splitlines()
        anzahl_zeilen = len(zeilen) - 1 # -1 weil erste Zeile Spaltennamen sind
        
    # print(f"Anzahl der Zeilen in der CSV-Datei: {anzahl_zeilen}") Anzahl der Spalten um Antworten zu überprüfen


    print("Aufgabe 1: Extract all order numbers from the text.")

    print() # Zeilenumbruch

    #regex_aufgabe_1 = r".*" einfache Variante
    regex_aufgabe_1 = r"^[A-Z0-9-]+$" # Spezificher, matcht einen String der aus Großbuchstaben, Zahlen und - besteht!
    
    order_numbers_in_csv = []
    
    for zeile in zeilen[1:]:
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_numbers = spalten[1] # Nimmt die zweite Spalte, was die order numbers enthält
        nummer = re.findall(regex_aufgabe_1, order_numbers) # Findet Inhalt der jeweiligen Zeile in der Spalte order numbers mit dem RegEx Befehl
        order_numbers_in_csv.append(nummer[0]) # Nimmt erstes Element in der Liste 

    print(order_numbers_in_csv) # Resultat printen
    if len(order_numbers_in_csv) == 100: # Resultat auf Anzahl der Zeilen prüfen
        print() # Zeilenumbruch
        print("Alle order numbers gefunden!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 2: Extract all product codes.")

    print() # Zeilenumbruch

    #regex_aufgabe_2 = r".*" einfache Variante
    regex_aufgabe_2 = r"^[A-Z0-9]+$" # Spezificher, matcht einen String der aus Großbuchstaben und Zahlen besteht!
    
    product_codes_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        product_code = spalten[0] # Nimmt die erste Spalte, was die product codes enthält
        codes = re.findall(regex_aufgabe_2, product_code) # Findet Inhalt der jeweiligen Zeile in der Spalte prodcut codes mit dem RegEx Befehl
        product_codes_in_csv.append(codes[0]) # Nimmt erstes Element in der Liste 

    print(product_codes_in_csv) # Resultat printen
    if len(product_codes_in_csv) == 100: # Resultat auf Anzahl der Zeilen prüfen
        print() # Zeilenumbruch
        print("Alle product codes gefunden!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 3: Extract all prices.")

    print() # Zeilenumbruch

    #regex_aufgabe_3 = r".*" einfache Variante
    regex_aufgabe_3 = r"^\$\d+\.\d+$" # Spezificher, matcht einen String der mit $ Zeichen beginnt, Zahlen folgen, ein Punkt und wieder Zahlen
    
    prices_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        preis = spalten[2] # Nimmt die dritte Spalte, die die Preise enthält
        preise = re.findall(regex_aufgabe_3, preis) # Findet Inhalt der jeweiligen Zeile in der Spalte price mit dem RegEx Befehl
        prices_in_csv.append(preise[0]) # Nimmt erstes Element in der Liste 

    print(prices_in_csv) # Resultat printen
    if len(prices_in_csv) == 100: # Resultat auf Anzahl der Zeilen prüfen
        print() # Zeilenumbruch
        print("Alle Preise gefunden!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 4: Extract all order dates.")

    print() # Zeilenumbruch

    #regex_aufgabe_4 = r".*" einfache Variante
    regex_aufgabe_4 = r"^[0-9/]+$" # Spezificher, matcht einen String der nur aus Zahlen und / Zeichen besteht!
    
    order_dates_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_date = spalten[4] # Nimmt die fünfte Spalte, die die Datum Angaben enthält
        order_dates = re.findall(regex_aufgabe_4, order_date) # Findet Inhalt der jeweiligen Zeile in der Spalte order date mit dem RegEx Befehl
        order_dates_in_csv.append(order_dates[0]) # Nimmt erstes Element in der Liste 

    print(order_dates_in_csv) # Resultat printen
    if len(order_dates_in_csv) == 100: # Resultat auf Anzahl der Zeilen prüfen
        print() # Zeilenumbruch
        print("Alle order dates gefunden!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 5: Find all orders for products priced over $500. (You are allowed to use Python to filter the list.)")
    
    print() # Zeilenumbruch

    #regex_aufgabe_5 = r".*" einfache Variante
    regex_aufgabe_5 = r"^\$\d+\.\d+$" # Spezificher, matcht einen String der nur aus Zahlen und / Zeichen besteht!
    
    order_prices_over_500_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_price_over_500 = spalten[2] # Nimmt die dritte Spalte, die die Preise enthält
        order_prices_over_500 = re.findall(regex_aufgabe_5, order_price_over_500) # Findet Inhalt der jeweiligen Zeile in der Spalte price mit dem RegEx Befehl
        price_erstes_element = order_prices_over_500[0] # Nimmt erstes Element der Liste
        order_prices_over_500_remove_dollar_sign = float(price_erstes_element.replace("$", "")) # Entfernt Dollar Zeichen
        if order_prices_over_500_remove_dollar_sign > 500:
            order_prices_over_500_in_csv.append(order_prices_over_500[0]) # Nimmt erstes Element in der Liste 

    orders_over_500 = []

    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        preis = spalten[2] # Nimmt die dritte Spalte, die die Preise enthält
        preis_ohne_dollar_sign = float(preis.replace("$", "")) # Entfernt Dollar Zeichen
        order_nummer = spalten[1]
        if preis_ohne_dollar_sign > 500:
            orders_over_500.append(order_nummer)

    print(orders_over_500) # Resultat printen
    print() # Zeilenumbruch
    print("Alle orders mit Produkten die insgesamt über 500$ kosten!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 6: Change the date format to DD/MM/YYYY. (Note the re.sub() method)")

    print() # Zeilenumbruch

    #regex_aufgabe_4 = r".*" einfache Variante
    regex_aufgabe_4 = r"^[0-9/]+$" # Spezificher, matcht einen String der nur aus Zahlen und / Zeichen besteht!
    
    order_dates_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_date = spalten[4] # Nimmt die fünfte Spalte, die die Datum Angaben enthält
        order_dates = re.findall(regex_aufgabe_4, order_date) # Findet Inhalt der jeweiligen Zeile in der Spalte order date mit dem RegEx Befehl
        order_dates_reihenfolge = re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\2/\1/\3", order_dates[0]) # re.sub Methode um Monat und Tag zu switchen
        order_dates_in_csv.append(order_dates_reihenfolge) # Fügt Datum mit richtiger Reihenfolge in die Liste

    print(order_dates_in_csv) # Resultat printen
    if len(order_dates_in_csv) == 100: # Resultat auf Anzahl der Zeilen prüfen
        print() # Zeilenumbruch
        print("Alle order dates in richtiger Reihenfolge gefunden!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 7: Find all orders with the highest number of ordered items.")

    print() # Zeilenumbruch

    #regex_aufgabe_7 = r".*" einfache Variante
    regex_aufgabe_7 = r"[0-9]+$" # Spezificher, matcht einen String der mit ein oder mehrere Zahlen von 0 - 9 hat
    
    number_of_items_in_csv = []
    number_of_items_in_csv_maximum = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        numbers_item = spalten[3] # Nimmt die vierte Spalte, die die number of items enthält
        numbers_items = re.findall(regex_aufgabe_7, numbers_item) # Findet Inhalt der jeweiligen Zeile in der Spalte price mit dem RegEx Befehl
        number_of_items_in_csv.append(numbers_items[0]) # Nimmt erstes Element in der Liste 
        number_of_items_in_csv_maximum.append(int(numbers_items[0])) # Nimmt erstes Element in der Liste als int 

        max_item = number_of_items_in_csv_maximum[0] # Mit erstem Wert in der Liste starten
        for itemnummer in number_of_items_in_csv_maximum: # Maximaler Item Wert finden
            if itemnummer > max_item:
                max_item = itemnummer
    print(f"{max_item} sind die maximale Anzahl an Produkten in den orders!") # Maximale Itemanzahl in den orders ist 10! 
    print() # Zeilenumbruch

    order_mit_10_items = []

    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        items = int(spalten[3]) # Nimmt die vierte Spalte, die die number of items enthält als int
        order_nummer = spalten[1]
        if items == max_item:
            order_mit_10_items.append(order_nummer)

    print(order_mit_10_items) # Resultat printen
    print() # Zeilenumbruch
    print("Alle orders die die maximale Anzahl an Produkten (10) haben!")

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch

    print("Aufgabe 8: Find the cheapest order(s). (You may want to use Python's min() method.)")

    print() # Zeilenumbruch

    #regex_aufgabe_8 = r".*" einfache Variante
    regex_aufgabe_8 = r"^\$\d+\.\d+$" # Spezificher, matcht einen String der nur aus Zahlen und / Zeichen besteht!
    
    order_prices_minimum_in_csv = []
    
    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_price_minimum = spalten[2] # Nimmt die dritte Spalte, die die Preise enthält
        order_prices_minimum = re.findall(regex_aufgabe_8, order_price_minimum) # Findet Inhalt der jeweiligen Zeile in der Spalte price mit dem RegEx Befehl
        order_prices_minimum_erstes_element = order_prices_minimum[0] # Nimmt erstes Element der Liste
        order_prices_minimum_erstes_element_remove_dollar_sign = float(order_prices_minimum_erstes_element.replace("$", "")) # Entfernt Dollar Zeichen
        order_prices_minimum_in_csv.append(order_prices_minimum_erstes_element_remove_dollar_sign)
        minimum_preis = min(order_prices_minimum_in_csv)

    orders_mit_min_preis = []

    for zeile in zeilen[1:]: # Erste Zeile überspringen weil sind nur die Spaltennamen
        spalten = zeile.split(",") # Die Spalten auftrennen
        order_nummer = spalten[1] # Nimmt die zweite Spalte, die die order Nummer enthält
        preis_der_order = float(spalten[2].replace("$",""))  # Preis als float, ohne Dollar Zeichen
        if preis_der_order == minimum_preis:
            orders_mit_min_preis.append(order_nummer)
        
    print(f"Die order mit dem geringsten Preis ist {orders_mit_min_preis[0]} und beträgt {minimum_preis}$!") # Resultat printen

    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch
    print() # Zeilenumbruch


if __name__ == "__main__":
    main()
```
